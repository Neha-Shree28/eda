```python
import pandas as pd
from scipy.stats import ttest_ind

df = pd.read_csv("student_data.csv")

male = df[df["sex"] == "M"]["G3"]
female = df[df["sex"] == "F"]["G3"]

t_value, p_value = ttest_ind(male, female)

alpha = 0.05

print("t-value :", t_value)
print("p-value :", p_value)

if p_value < alpha:
    print("\nConclusion:")
    print("Reject the Null Hypothesis (H0)")
    print("There is a significant difference between")
    print("the G3 scores of male and female students.")
else:
    print("\nConclusion:")
    print("Fail to Reject the Null Hypothesis (H0)")
    print("There is no significant difference between")
    print("the G3 scores of male and female students.")
```
