```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("student_data.csv")

# Create Pair Plot
sns.pairplot(
    df,
    vars=["age", "studytime", "failures", "absences", "G1", "G2", "G3"],
    hue="sex",
    diag_kind="hist"
)

# Display the plot
plt.show()
```
