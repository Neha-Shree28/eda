```python
import pandas as pd

df = pd.read_csv("student_data.csv")

df["Rolling_Mean"] = df["G3"].rolling(window=7).mean()

df["Rolling_SD"] = df["G3"].rolling(window=7).std()

print(df[["G3", "Rolling_Mean", "Rolling_SD"]])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("student_data.csv")

df["Rolling_Mean"] = df["G3"].rolling(window=7).mean()

plt.figure(figsize=(10, 5))

plt.plot(df.index, df["G3"], label="Original G3")
plt.plot(df.index, df["Rolling_Mean"],
         linewidth=3, label="7-Student Rolling Mean")

plt.title("Student Final Grade vs 7-Student Rolling Mean")
plt.xlabel("Student Index")
plt.ylabel("Final Grade (G3)")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("student_data.csv")

# Calculate 7-student Rolling Standard Deviation
df["Rolling_SD"] = df["G3"].rolling(window=7).std()

# Plot
plt.figure(figsize=(10, 5))

plt.plot(df.index, df["Rolling_SD"],
         linewidth=3)

plt.title("7-Student Rolling Standard Deviation of G3")
plt.xlabel("Student Index")
plt.ylabel("Standard Deviation")

plt.show()
```
